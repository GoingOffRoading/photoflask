```python
import os
import sqlite3
import uuid
from datetime import datetime
```

```python
def db_path_function(db_filename="photoflask.db"):
   # Checks to see if DB file exists, and if not, creates it. Returns the path to the DB file.

    base_dir = os.path.dirname(os.path.abspath(__file__)) if "__file__" in globals() else os.getcwd()
    db_path = os.path.join(base_dir, db_filename)
    if not os.path.exists(db_path):
        print(f"Creating database file at {db_path}...")
        sqlite3.connect(db_path).close()
    else:
        print(f"Database file already exists at {db_path}")
    return db_path
```

```python
def ensure_photos_table(db_path=db_path_function()):
    # Checks to see if a Photos table exists, and if not, creates one with:
    # - Guid (primary key)
    # - FileName
    # - FilePath
    # - MediaType (image or video)
    # - LastTouched

    conn = sqlite3.connect(db_path)
    cursor = conn.cursor()
    
    # Check if the 'photos' table exists
    cursor.execute("""
        SELECT name FROM sqlite_master 
        WHERE type='table' AND name='photos'
    """)
    
    if cursor.fetchone() is None:
        # Table doesn't exist, create it
        print(f"'photos' table does not exist in {db_path}. Creating table...")
        cursor.execute("""
            CREATE TABLE photos (
                Guid TEXT PRIMARY KEY,
                FileName TEXT NOT NULL,
                FilePath TEXT NOT NULL,
                MediaType TEXT DEFAULT 'unknown',
                LastTouched TIMESTAMP DEFAULT CURRENT_TIMESTAMP
            )
        """)
        conn.commit()
        print(f"Created 'photos' table in {db_path}")
    else:
        # Table exists, check if MediaType column exists
        cursor.execute("PRAGMA table_info(photos)")
        columns = [col[1] for col in cursor.fetchall()]
        
        if 'MediaType' not in columns:
            print(f"Adding 'MediaType' column to 'photos' table...")
            cursor.execute("""
                ALTER TABLE photos
                ADD COLUMN MediaType TEXT DEFAULT 'unknown'
            """)
            conn.commit()
            print(f"Added 'MediaType' column to 'photos' table in {db_path}")
        else:
            print(f"'photos' table already exists in {db_path}")
    
    conn.close()
```

```python
def delete_database(db_path=db_path_function()):
    # Function to delete photoflask.db
    
    if os.path.exists(db_path):
        os.remove(db_path)
        print(f"Deleted database file: {db_path}")
    else:
        print(f"Database file not found: {db_path}")
```

```python
# To create/ensure the photos table exists
ensure_photos_table()
```

```python
delete_database()
```
